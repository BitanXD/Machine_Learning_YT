```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
documents = [
    'I love pizza',
    'Pizza is the best',
    'I love pasta',
    'Pasta is great'
]
```

```python
vetorise = CountVectorizer(ngram_range=(2,2))
```

```python
X = vetorise.fit_transform(documents)
```

```python
print('Vocabulary: ', vetorise.get_feature_names_out())
```

```python
print('BoW Matrix: \n', X.toarray())
```

```python
vectorise = TfidfVectorizer()
```

```python
X = vectorise.fit_transform(documents)
```

```python
print('Vocabulary: ', vetorise.get_feature_names_out())
```

```python
print('BoW Matrix: \n', X.toarray())
```
