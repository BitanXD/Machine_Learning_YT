```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('data/train.txt', sep=';',header=None, names=['text','emotion'])
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
unique_emotions = df['emotion'].unique()
```

```python
emotions_dict = {}
i = 0
for emotion in unique_emotions:
    emotions_dict[emotion] = i
    i += 1

df['emotion'] = df['emotion'].map(emotions_dict)
df.head()
```

```python
# convert to lower case
df['text'] = df['text'].apply(lambda x : x.lower())
```

```python
#  remove punctutations
import string

def remove_punc(txt):
    return txt.translate(str.maketrans('','',string.punctuation))
```

```python
df['text'] = df['text'].apply(remove_punc)
df.head()
```

```python
#remove  numbers
def remove_numbers(txt):
    new_txt =''
    for i in txt:
        if not i.isdigit():
            new_txt = new_txt + i
    return new_txt
```

```python
df['text'] = df['text'].apply(remove_numbers)
df.head()
```

```python
# remove emojis and special characters
def remove_spcl(txt):
    new_txt =''
    for i in txt:
        if i.isascii():
            new_txt = new_txt + i
    return new_txt
```

```python
df['text'] = df['text'].apply(remove_spcl)
df.head()
```

```python
import nltk
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab')
```

```python
stop_words = set(stopwords.words('english'))
len(stop_words)
```

```python
# Removing stopwords
def  remove_stopwords(txt):
    words = word_tokenize(txt)
    clean_txt = []
    for w in words:
        if not w in stop_words:
            clean_txt.append(w)
    return ' '.join(clean_txt)
```

```python
df['text']= df['text'].apply(remove_stopwords)
df.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(df['text'], df['emotion'], test_size=0.20, random_state=42)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.feature_extraction.text import CountVectorizer
```

```python
bow_vectoizer = CountVectorizer()
```

```python
X_train_bow = bow_vectoizer.fit_transform(X_train)
X_test_bow = bow_vectoizer.transform(X_test)
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
```

```python
nb_model = MultinomialNB()
nb_model.fit(X_train_bow,y_train)
```

```python
y_pred = nb_model.predict(X_test_bow)
accuracy_score(y_test, y_pred)
```

```python
tf_idf_vectorizer = TfidfVectorizer()
```

```python
X_train_tf = tf_idf_vectorizer.fit_transform(X_train)
X_test_tf = tf_idf_vectorizer.transform(X_test)
```

```python
nb_model2 = MultinomialNB()
```

```python
nb_model2.fit(X_train_tf, y_train)
```

```python
y_pred = nb_model2.predict(X_test_tf)
accuracy_score(y_test,y_pred)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr_model = LogisticRegression(max_iter=1000)
```

```python
lr_model.fit(X_train_tf, y_train)
```

```python
y_pred = lr_model.predict(X_test_tf)
```

```python
accuracy_score(y_test,y_pred)
```
