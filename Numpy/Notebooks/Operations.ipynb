```python
import numpy as np
```

### Arithmatic Operations

```python
a1 = np.arange(1,6)
a2 = np.arange(6,11)
```

```python
result_sum = a1 + a2
result_sum
```

```python
result_remainder = a2 // a1
result_remainder
```

### Broadcasting

```python
l = [10,20,30,40,50,60]
arr_l = np.array(l)
```

```python
arr_l + 5
```

### Deep and Shallow Copy

```python
arr_ds = np.arange(1,26).reshape(5,5)
arr_ds
```

```python
arr_ds + 10
```

```python
arr_ds
```

```python
arr = np.arange(1,11)
```

```python
arr_slice = arr[: 5]
arr_slice = arr_slice * 10
arr_slice
```

```python
arr # this is shallow copy where changes made in slice of array does not affect the main array
```

```python
brr = arr
brr[0] = 999
```

```python
brr
```

```python
arr # this is deep copy where changes made are reflected in main array as well because both point to same memory location
```

```python
brr_copy = arr.copy()
brr_copy[0] = 111
```

```python
brr_copy
```

```python
arr
```

### Matrix Operations

```python
A = np.array([[1,2],
            [3,4]])
B = np.array([[5,6],
            [7,8]])
```

```python
A @ B
```

```python
A.T
```

### Stacking Array

```python
a = np.arange(1,6)
b = np.arange(6,11)
```

```python
np.vstack((a,b))
```

```python
np.hstack((a,b))
```

```python
np.column_stack((a,b))
```
