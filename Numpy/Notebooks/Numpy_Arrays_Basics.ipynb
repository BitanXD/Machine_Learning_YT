## Reshaping and Resizing

```python
arr = np.arange(1,31)
arr
```

```python
arr.shape
```

```python
arr = arr.reshape(6,5) # resize
arr
```

```python
arr.shape
```

```python
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
arr[0,0] # same as arr[0][0]
```

```python
slice = arr[0:2,1:3] 
slice
```

## Boolean Index

```python
arr = np.arange(1,11)
arr
```

```python
boolean_index = arr % 2 == 0
boolean_index
```

```python
arr_bool = arr[boolean_index]
arr_bool
```
